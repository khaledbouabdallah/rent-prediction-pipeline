# GEO PLOTS

```python


# Create a GeoDataFrame
gdf = gpd.GeoDataFrame(
    df, geometry=gpd.points_from_xy(df['longitude'], df['latitude'])
)

# Load the geographical data for French departments
departments = gpd.read_file("https://raw.githubusercontent.com/gregoiredavid/france-geojson/master/departements-version-simplifiee.geojson")
```

```python
mean_price_by_dept = df.groupby('department_number')['price'].mean().reset_index()

departments_with_prices = departments.merge(
    mean_price_by_dept, left_on='code', right_on='department_number', how='left'
)
```
